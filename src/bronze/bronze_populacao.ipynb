```python
from pyspark.sql.types import *
import pyspark.sql.functions as F
```

```python
schema_populacao = StructType([
    StructField("uf", StringType(), nullable= False),	
    StructField("codigo_uf", StringType(), nullable= False),	
    StructField("codigo_municipio", StringType(), nullable= False),	
    StructField("nome_municipio", StringType(), nullable= False),	
    StructField("populacao_estimada", StringType(), nullable= False)
])
```

```python
df = spark.read.csv("/Volumes/workspace/businessradar/raw_data/populacao_ibge/", header =False, encoding= "UTF-8", schema= schema_populacao, sep=";")
```

```python
df.count()
```

```python
df.show()
```

```python
df_populacao = (df.withColumn("arquivo_origem", F.col("_metadata.file_name")))
```

```python
df_populacao.show(5)
```

```python
df_populacao.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("workspace.businessradar.bronze_populacao")
```

```python
df_populacao_read = spark.read.table("workspace.businessradar.bronze_populacao")
```
