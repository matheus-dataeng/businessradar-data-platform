```python
from pyspark.sql.types import *
import pyspark.sql.functions as F
```

```python
schema_municipios = StructType([
    StructField("codigo_municipio", StringType(), nullable=False),
    StructField("codigo_ibge", StringType(), nullable=False),
    StructField("nome_municipio_tom", StringType(), nullable=False),
    StructField("nome_municipio_ibge", StringType(), nullable=True),
    StructField("uf", StringType(), nullable=False),
])
```

```python
df = spark.read.csv("/Volumes/workspace/businessradar/raw_data/municipios/", header= True,encoding= "ISO-8859-1", sep= ";", schema= schema_municipios)
```

```python
df.count()
```

```python
df_municipios = (df.withColumn("arquivo_origem", F.col("_metadata.file_name")))
```

```python
df_municipios.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("workspace.businessradar.bronze_municipios")
```

```python
df_municipios_read = spark.read.table("workspace.businessradar.bronze_municipios")
```
