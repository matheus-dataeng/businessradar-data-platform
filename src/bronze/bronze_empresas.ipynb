```python
from pyspark.sql.types import *
import pyspark.sql.functions as F
```

```python
schema_empresas = StructType([
    StructField("cnpj_basico", StringType(), nullable= True),
    StructField("razao_social", StringType(), nullable= True),
    StructField("natureza_juridica", StringType(), nullable= True),
    StructField("qualificacao_responsavel", StringType(), nullable= True),
    StructField("capital_social", StringType(), nullable= True),
    StructField("porte_empresa", StringType(), nullable= True),
    StructField("ente_federativo_responsavel", StringType(), nullable= True)
])
```

```python
df = spark.read.csv("/Volumes/workspace/businessradar/raw_data/empresas/", schema= schema_empresas, header= False, sep= ";", escape= '"', encoding= "ISO-8859-1")
```

```python
df.show()
```

```python
df.printSchema()
```

```python
df_empresas = (
    df.withColumn("arquivo_origem", F.col("_metadata.file_name"))
)
```

```python
df_empresas.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("workspace.businessradar.bronze_empresas")
```

```python
df_empresas_read = spark.read.table("workspace.businessradar.bronze_empresas")
```
