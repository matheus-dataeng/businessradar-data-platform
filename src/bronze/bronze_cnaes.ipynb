```python
from pyspark.sql.types import *
import pyspark.sql.functions as F
```

```python
schema_cnaes = StructType([
    StructField("codigo_cnae", StringType(), nullable= False), 
    StructField("descricao_cnae", StringType(), nullable= False) 
])
```

```python
df = spark.read.csv("/Volumes/workspace/businessradar/raw_data/cnaes/", schema= schema_cnaes, encoding= "ISO-8859-1", sep= ";")
```

```python
df.printSchema()
```

```python
df.count()
```

```python
df_cnaes = (df.withColumn("arquivo_origem", F.col("_metadata.file_name")))
```

```python
df_cnaes.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("workspace.businessradar.bronze_cnaes") 
```

```python
df_cnaes_read = spark.read.table("workspace.businessradar.bronze_cnaes")
```
