```python
from pyspark.sql.types import *
import pyspark.sql.functions as F
```

```python
schema_simples = StructType([
    StructField("cnpj_basico", StringType()),	
	StructField("opcao_pelo_simples", StringType()),	
	StructField("data_opcao_simples", StringType()),	
	StructField("data_exclusao_simples", StringType()),	
	StructField("opcao_pelo_mei", StringType()),	
	StructField("data_opcao_mei", StringType()),	
	StructField("data_exclusao_mei", StringType())
])
```

```python
df = spark.read.csv("/Volumes/workspace/businessradar/raw_data/simples/", header= False, encoding= "ISO-8859-1", schema= schema_simples, sep= ";")
```

```python
df.count()
```

```python
df.printSchema()
```

```python
df_simples = (df.withColumn("arquivo_origem", F.col("_metadata.file_name")))
```

```python
df_simples.show(5)
```

```python
df_simples.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("workspace.businessradar.bronze_simples")
```

```python
df_simples_read = spark.read.table("workspace.businessradar.bronze_simples")
```
