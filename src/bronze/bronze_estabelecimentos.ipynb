```python
from pyspark.sql.types import *
import pyspark.sql.functions as F
```

```python
schema_estabelecimentos = StructType([
    StructField("cnpj_basico", StringType(), nullable= True),
	StructField("cnpj_ordem", StringType(), nullable= True),
	StructField("cnpj_dv", StringType(), nullable= True),
	StructField("identificador_matriz_filial", StringType(), nullable= True),
	StructField("nome_fantasia", StringType(), nullable= True),
	StructField("situacao_cadastral", StringType(), nullable= True),
	StructField("data_situacao_cadastral", StringType(), nullable= True),
	StructField("motivo_situacao_cadastral", StringType(), nullable= True),
	StructField("nome_cidade_exterior", StringType(), nullable= True),
	StructField("pais", StringType(), nullable= True),
	StructField("data_inicio_atividade", StringType(), nullable= True),
	StructField("cnae_fiscal_principal", StringType(), nullable= True),
	StructField("cnae_fiscal_secundaria", StringType(), nullable= True),
	StructField("tipo_logradouro", StringType(), nullable= True),
	StructField("logradouro", StringType(), nullable= True),
	StructField("numero", StringType(), nullable= True),
	StructField("complemento", StringType(), nullable= True),
	StructField("bairro", StringType(), nullable= True),
	StructField("cep", StringType(), nullable= True),
	StructField("uf", StringType(), nullable= True),
	StructField("municipio", StringType(), nullable= True),
	StructField("ddd_1", StringType(), nullable= True),
	StructField("telefone_1", StringType(), nullable= True),
	StructField("ddd_2", StringType(), nullable= True),
	StructField("telefone_2", StringType(), nullable= True),
	StructField("ddd_fax", StringType(), nullable= True),
	StructField("fax", StringType(), nullable= True),
	StructField("correio_eletronico", StringType(), nullable= True),
	StructField("situacao_especial", StringType(), nullable= True),
	StructField("data_situacao_especial", StringType(), nullable= True)
])	
```

```python
df = spark.read.option("multiLine", "true").csv("/Volumes/workspace/businessradar/raw_data/estabelecimentos /", encoding= "ISO-8859-1", header= False, sep= ";", escape= '"', schema= schema_estabelecimentos)
```

```python
df.show(1)
```

```python
df_estabelecimentos = (
    df.withColumn("arquivo_origem", F.col("_metadata.file_name"))
)
```

```python
df_estabelecimentos.show(5)
```

```python
df_estabelecimentos.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("workspace.businessradar.bronze_estabelecimentos")
```

```python
df_estabelecimentos_read = spark.read.table("workspace.businessradar.bronze_estabelecimentos")
```
